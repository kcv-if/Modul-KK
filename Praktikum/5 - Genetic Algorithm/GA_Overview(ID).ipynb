# ALGORITMA GENETIKA

Algoritma Genetika (Genetic Algorithm / GA) terinspirasi dari proses evolusi alami dan sangat berguna untuk menyelesaikan masalah optimasi dan pencarian dalam ruang pencarian yang sangat besar.

Dalam menyelesaikan suatu masalah, algoritma ini menggunakan *populasi* solusi (juga disebut *state* atau keadaan), di mana setiap solusi/state merepresentasikan solusi yang layak. Pada setiap iterasi (sering disebut *generasi*), populasi diperbarui menggunakan metode yang terinspirasi dari biologi dan evolusi, seperti *crossover* (persilangan), *mutasi*, dan *seleksi alam*.

## Gambaran Umum

Algoritma genetika bekerja dengan cara berikut:

1) **Inisialisasi populasi acak** - Membuat populasi awal dengan solusi-solusi acak.

2) **Hitung fitness populasi** - Menghitung seberapa baik setiap solusi (fitness score).

3) **Seleksi individu untuk kawin** - Memilih individu-individu terbaik untuk dijadikan orang tua.

4) **Kawinkan individu terpilih untuk menghasilkan populasi baru** - Menggabungkan gen dari orang tua.
     * Kesempatan acak untuk mutasi individu 

5) **Ulangi dari langkah 2)** - Terus lakukan sampai ditemukan solusi yang cukup baik atau mencapai jumlah iterasi maksimum.

## Istilah Penting

Sebelum melanjutkan, mari kita pahami istilah-istilah dasar dalam algoritma ini:

* **Individu/State**: Sebuah daftar elemen (disebut *gen*) yang merepresentasikan kemungkinan solusi.

* **Populasi**: Kumpulan dari semua individu/state.

* **Gene pool (Kumpulan Gen)**: Alfabet atau kumpulan nilai-nilai yang mungkin untuk gen suatu individu.

* **Generasi/Iterasi**: Jumlah berapa kali populasi akan diperbarui.

* **Fitness (Kebugaran)**: Skor individu, dihitung oleh fungsi fitness yang spesifik untuk masalah yang dihadapi.

## Crossover (Persilangan)

Dua individu/state dapat "kawin" dan menghasilkan satu anak. Keturunan ini mewarisi karakteristik dari kedua orang tuanya. Ada banyak cara untuk mengimplementasikan crossover ini. Di sini kita akan melihat yang paling umum:

* **Point Crossover**: Persilangan terjadi di sekitar satu titik (atau lebih). Orang tua "dipotong" pada titik yang dipilih kemudian digabungkan. Pada contoh di bawah, kita melihat dua orang tua dipotong dan digabungkan pada digit ke-3:

![point crossover](https://github.com/aimacode/aima-python/blob/master/images/point_crossover.png?raw=1)

* **Uniform Crossover**: Jenis crossover ini memilih gen secara acak untuk digabungkan. Di sini gen 1, 2, dan 5 dipilih dari orang tua pertama, jadi gen 3 dan 4 ditambahkan dari orang tua kedua:

![uniform crossover](https://github.com/aimacode/aima-python/blob/master/images/uniform_crossover.png?raw=1)

## Mutasi

Ketika keturunan dihasilkan, ada kemungkinan ia akan bermutasi, yaitu satu gen (atau lebih, tergantung implementasi) nya berubah secara acak.

Misalnya, individu baru yang akan mengalami mutasi adalah "abcde". Secara acak kita memilih untuk mengubah gen ketiga menjadi 'z'. Individu sekarang menjadi "abzde" dan ditambahkan ke populasi.

## Seleksi

Pada setiap iterasi, individu yang paling fit (terbaik) dipilih secara acak untuk kawin dan menghasilkan keturunan. Kita mengukur fitness individu dengan *fungsi fitness*. Fungsi itu bergantung pada masalah yang diberikan dan digunakan untuk memberi skor pada individu. Biasanya semakin tinggi skornya semakin baik.

Proses seleksi adalah sebagai berikut:

1) Individu diberi skor oleh fungsi fitness.

2) Individu dipilih secara acak, sesuai dengan skornya (skor lebih tinggi berarti peluang lebih besar untuk dipilih). Biasanya rumus untuk menghitung peluang memilih individu adalah sebagai berikut (untuk populasi *P* dan individu *i*):

$$ peluang(i) = \dfrac{fitness(i)}{\sum_{k \, dalam \, P}{fitness(k)}} $$

### Pendekatan dengan Algoritma Genetika

Sekarang kita akan menerapkan algoritma genetika dan melihat bagaimana ia secara signifikan mengurangi ruang pencarian. Kita pada dasarnya ingin *mengevolusi* populasi string acak kita sehingga mereka lebih mendekati frasa target seiring bertambahnya generasi.

Algoritma genetika bekerja berdasarkan prinsip Seleksi Alam Darwin yang memiliki tiga konsep kunci:

* **Hereditas (Keturunan)**: Harus ada proses di mana anak menerima sifat dari orang tuanya.  
Untuk masalah ini, dua string dari populasi akan dipilih sebagai orang tua dan akan dipotong pada indeks acak lalu digabungkan kembali seperti yang dijelaskan dalam fungsi `recombine` untuk membuat anak. String anak ini kemudian akan ditambahkan ke generasi baru.

* **Variasi**: Harus ada berbagai sifat yang ada dalam populasi atau cara untuk memperkenalkan variasi.  
Jika tidak ada variasi dalam ruang sampel, kita mungkin tidak akan pernah mencapai optimum global. Untuk memastikan ada cukup variasi, kita dapat menginisialisasi populasi besar, tetapi ini menjadi mahal secara komputasi. Oleh karena itu, kita sering menggunakan metode lain yang disebut mutasi. Dalam metode ini, kita secara acak mengubah satu atau lebih karakter dari beberapa string dalam populasi berdasarkan nilai probabilitas yang telah ditentukan. Tingkat mutasi biasanya dijaga cukup rendah (biasanya 5-10%).

* **Seleksi**: Harus ada mekanisme di mana beberapa anggota populasi memiliki kesempatan untuk menjadi orang tua dan meneruskan informasi genetiknya sementara yang lain tidak. Ini biasanya disebut sebagai "survival of the fittest".  
Harus ada cara untuk menentukan frasa mana dalam populasi kita yang memiliki peluang lebih baik untuk akhirnya berevolusi menjadi frasa target. Ini dilakukan dengan memperkenalkan fungsi fitness yang menghitung seberapa dekat frasa yang dihasilkan dengan frasa target. Fungsi akan mengembalikan nilai skalar yang sesuai dengan jumlah karakter yang cocok antara frasa yang dihasilkan dan frasa target.

### Langkah 1: Definisikan Frasa Target

Sebelum menyelesaikan masalah, kita perlu mendefinisikan frasa target kita:

```python
target = 'Genetic Algorithm'
```

### Langkah 2: Definisikan Gene Pool

Kemudian kita perlu mendefinisikan gene pool kita, yaitu elemen yang mungkin dimiliki oleh individu dari populasi. Di sini, gene pool berisi semua huruf besar dan kecil dari alfabet Inggris serta karakter spasi:

```python
# Nilai ASCII huruf besar berkisar dari 65 hingga 91
u_case = [chr(x) for x in range(65, 91)]
# Nilai ASCII huruf kecil berkisar dari 97 hingga 123
l_case = [chr(x) for x in range(97, 123)]

gene_pool = []
gene_pool.extend(u_case)  # menambahkan huruf besar ke gene pool
gene_pool.extend(l_case)  # menambahkan huruf kecil ke gene pool
gene_pool.append(' ')     # menambahkan karakter spasi ke gene pool

print(f"Gene pool: {gene_pool}")
print(f"Ukuran gene pool: {len(gene_pool)}")
```

### Langkah 3: Tentukan Ukuran Populasi

Sekarang kita perlu menentukan ukuran maksimum setiap populasi. Populasi yang lebih besar memiliki lebih banyak variasi tetapi lebih mahal secara komputasi untuk menjalankan algoritma:

```python
max_population = 100
```

### Langkah 4: Tentukan Tingkat Mutasi

Karena populasi kita tidak terlalu besar, kita dapat mempertahankan tingkat mutasi yang relatif besar:

```python
mutation_rate = 0.07  # 7%
```

### Langkah 5: Definisikan Fungsi Fitness

Bagus! Sekarang, kita perlu mendefinisikan metrik yang paling penting untuk algoritma genetika, yaitu fungsi fitness. Ini akan mengembalikan jumlah karakter yang cocok antara sampel yang dihasilkan dan frasa target:

```python
def fitness_fn(sample):
    # inisialisasi fitness ke 0
    fitness = 0
    for i in range(len(sample)):
        # tambahkan fitness sebesar 1 untuk setiap karakter yang cocok
        if sample[i] == target[i]:
            fitness += 1
    return fitness
```

### Langkah 6: Inisialisasi Populasi

Sebelum kita menjalankan algoritma genetika, kita perlu menginisialisasi populasi acak. Kita akan menggunakan fungsi `init_population` untuk melakukan ini. Kita perlu memasukkan ukuran populasi maksimum, gene pool, dan panjang setiap individu, yang dalam hal ini akan sama dengan panjang frasa target:

```python
population = init_population(max_population, gene_pool, len(target))
print(f"Contoh 5 individu pertama dari populasi awal:")
for i in range(5):
    print(f"Individu {i+1}: {''.join(population[i])}")
```

### Langkah 7: Memahami Proses Evolusi

Sekarang kita akan mendefinisikan bagaimana individu dalam populasi harus berubah seiring bertambahnya generasi.

Pertama, fungsi `select` akan dijalankan pada populasi untuk memilih *dua* individu dengan nilai fitness tinggi. Ini akan menjadi orang tua yang kemudian akan digabungkan menggunakan fungsi `recombine` untuk menghasilkan anak:

```python
# Pilih 2 orang tua
parents = select(2, population, fitness_fn)
print(f"Orang tua terpilih:")
print(f"Parent 1: {''.join(parents[0])}")
print(f"Parent 2: {''.join(parents[1])}")
```

```python
# Fungsi recombine mengambil dua orang tua sebagai argumen, jadi kita perlu unpack variabel sebelumnya
child = recombine(*parents)
print(f"Anak hasil crossover: {''.join(child)}")
```

Selanjutnya, kita perlu menerapkan mutasi sesuai dengan tingkat mutasi. Kita memanggil fungsi `mutate` pada anak dengan gene pool dan tingkat mutasi sebagai argumen tambahan:

```python
child = mutate(child, gene_pool, mutation_rate)
print(f"Anak setelah mutasi: {''.join(child)}")
```

Baris di atas dapat dipadatkan menjadi:

`child = mutate(recombine(*select(2, population, fitness_fn)), gene_pool, mutation_rate)`

Dan, kita perlu melakukan ini `untuk` setiap individu dalam populasi saat ini untuk menghasilkan populasi baru:

```python
population = [mutate(recombine(*select(2, population, fitness_fn)), gene_pool, mutation_rate) 
              for i in range(len(population))]
```

Individu dengan fitness tertinggi kemudian dapat ditemukan menggunakan fungsi `max`:

```python
current_best = max(population, key=fitness_fn)
print(f"Individu terbaik saat ini: {current_best}")
```

Kita lihat bahwa ini adalah list karakter. Ini dapat dikonversi menjadi string menggunakan fungsi join:

```python
current_best_string = ''.join(current_best)
print(f"Frasa terbaik: {current_best_string}")
print(f"Fitness: {fitness_fn(current_best)}")
```

### Langkah 8: Definisikan Kondisi Terminasi

Sekarang kita perlu mendefinisikan kondisi untuk menghentikan algoritma. Ini dapat terjadi dalam dua cara:
1. Terminasi setelah jumlah generasi yang telah ditentukan
2. Terminasi ketika fitness individu terbaik dari generasi saat ini mencapai nilai threshold yang telah ditentukan

Kita definisikan variabel-variabel ini di bawah:

```python
ngen = 1200  # jumlah maksimum generasi
# kita set threshold fitness sama dengan panjang frasa target
# yaitu algoritma hanya berhenti ketika semua karakter sudah benar
# atau sudah menyelesaikan 'ngen' jumlah generasi
f_thres = len(target)
```

### Langkah 9: Membuat Fungsi untuk Menjalankan Algoritma

Untuk menghasilkan `ngen` jumlah generasi, kita menjalankan loop `for` sebanyak `ngen` kali. Setelah setiap generasi, kita menghitung fitness individu terbaik dari generasi tersebut dan membandingkannya dengan nilai `f_thres` menggunakan fungsi `fitness_threshold`. Setelah setiap generasi, kita mencetak individu terbaik dari generasi tersebut dan nilai fitness yang sesuai. Mari kita tulis fungsi untuk melakukan ini:

```python
def genetic_algorithm_stepwise(population, fitness_fn, gene_pool=[0, 1], f_thres=None, ngen=1200, pmut=0.1):
    """Menjalankan algoritma genetika dengan menampilkan progress setiap generasi."""
    for generation in range(ngen):
        # Buat populasi baru
        population = [mutate(recombine(*select(2, population, fitness_fn)), gene_pool, pmut) 
                      for i in range(len(population))]
        
        # Simpan genom individu dengan fitness tertinggi dalam populasi saat ini
        current_best = ''.join(max(population, key=fitness_fn))
        print(f'Terbaik saat ini: {current_best}\t\tGenerasi: {str(generation)}\t\tFitness: {fitness_fn(current_best)}\r', end='')

        # Bandingkan fitness individu terbaik saat ini dengan f_thres
        fittest_individual = fitness_threshold(fitness_fn, f_thres, population)

        # Jika fitness lebih besar atau sama dengan f_thres, kita hentikan algoritma
        if fittest_individual:
            return fittest_individual, generation
            
    return max(population, key=fitness_fn), generation
```

Fungsi yang didefinisikan di atas pada dasarnya sama dengan yang didefinisikan di `search.py` dengan fungsionalitas tambahan untuk mencetak data setiap generasi.

Mari kita lihat implementasi asli dari `genetic_algorithm`:

```python
psource(genetic_algorithm)
```

### Langkah 10: Jalankan Algoritma

Kita telah mendefinisikan semua fungsi dan variabel yang diperlukan. Sekarang mari kita buat populasi baru dan uji fungsi yang kita tulis di atas:

```python
# Buat populasi baru
population = init_population(max_population, gene_pool, len(target))

# Jalankan algoritma
solution, generations = genetic_algorithm_stepwise(population, fitness_fn, gene_pool, f_thres, ngen, mutation_rate)

print(f"\n\nSolusi ditemukan: {''.join(solution)}")
print(f"Jumlah generasi: {generations}")
print(f"Fitness akhir: {fitness_fn(solution)}")
```

## 1. Pewarnaan Graf (Graph Coloring)

Pertama kita akan menangani masalah yang lebih sederhana yaitu mewarnai graf kecil dengan dua warna. Sebelum kita melakukan apa pun, mari kita bayangkan bagaimana solusinya mungkin terlihat.

### Representasi Solusi

Pertama, kita harus merepresentasikan warna kita. Katakanlah, 'R' untuk merah dan 'G' untuk hijau. Ini membentuk gene pool kita. Bagaimana dengan solusi individu? Untuk itu, kita akan melihat masalah kita. Kita menyatakan kita memiliki graf. Sebuah graf memiliki node dan edge, dan kita ingin mewarnai node. Secara alami, kita ingin menyimpan warna setiap node. Jika kita memiliki empat node, kita dapat menyimpan warnanya dalam list gen, satu untuk setiap node. Solusi yang mungkin kemudian akan terlihat seperti ini: `['R', 'R', 'G', 'R']`. Dalam kasus umum, kita akan merepresentasikan setiap solusi dengan list karakter ('R' dan 'G'), dengan panjang jumlah node.

### Fungsi Fitness

Selanjutnya kita perlu membuat fungsi fitness yang memberi skor individu dengan tepat. Sekali lagi, kita akan melihat definisi masalah yang ada. Kita ingin mewarnai graf. Agar solusi menjadi optimal, tidak ada edge yang harus menghubungkan dua node dengan warna yang sama. Bagaimana kita dapat menggunakan informasi ini untuk memberi skor pada solusi?

Pendekatan naif (dan tidak efektif) adalah menghitung warna yang berbeda dalam string. Jadi `['R', 'R', 'R', 'R']` memiliki skor 1 dan `['R', 'R', 'G', 'G']` memiliki skor 2. Mengapa fungsi fitness itu tidak ideal? Karena kita lupa informasi tentang edge! Edge sangat penting untuk masalah dan fungsi di atas hanya berurusan dengan warna node.

Kita mengatakan bahwa solusi optimal akan memiliki semua edge yang menghubungkan node dengan warna berbeda. Jadi, untuk memberi skor pada solusi, kita dapat menghitung berapa banyak edge yang valid (alias menghubungkan node dengan warna berbeda). Itu adalah fungsi fitness yang bagus!

### Implementasi Graph Coloring

Pertama kita perlu merepresentasikan graf. Karena kita sebagian besar membutuhkan informasi tentang edge, kita hanya akan menyimpan edge. Kita akan menunjukkan edge dengan huruf kapital dan node dengan integer:

```python
edges = {
    'A': [0, 1],
    'B': [0, 3],
    'C': [1, 2],
    'D': [2, 3]
}

print("Graf yang akan diwarnai:")
print("Edge 'A' menghubungkan node 0 dan 1")
print("Edge 'B' menghubungkan node 0 dan 3")
print("Edge 'C' menghubungkan node 1 dan 2")
print("Edge 'D' menghubungkan node 2 dan 3")
```

Kita sudah mengatakan gene pool kita adalah 'R' dan 'G', jadi kita dapat langsung menginisialisasi populasi kita. Karena kita hanya memiliki empat node, `state_length` harus 4. Untuk jumlah individu, kita akan mencoba 8:

```python
population = init_population(8, ['R', 'G'], 4)
print("Populasi awal (8 individu dengan 4 gen masing-masing):")
for i, individual in enumerate(population):
    print(f"Individu {i+1}: {individual}")
```

Kita membuat dan mencetak populasi. Terlihat bahwa gen dalam individu adalah acak dan ada 8 individu masing-masing dengan 4 gen.

Selanjutnya kita perlu menulis fungsi fitness kita. Kita sebelumnya mengatakan kita ingin fungsi menghitung berapa banyak edge yang valid. Jadi, diberikan pewarnaan/individu `c`, kita akan melakukan hal itu:

```python
def fitness(c):
    """Menghitung jumlah edge yang valid (node berbeda warna)."""
    return sum(c[n1] != c[n2] for (n1, n2) in edges.values())
```

Bagus! Sekarang kita akan menjalankan algoritma genetika dan melihat solusi apa yang diberikannya:

```python
solution = genetic_algorithm(population, fitness, gene_pool=['R', 'G'])
print(f"Solusi: {solution}")
print(f"Fitness: {fitness(solution)}")
```

Algoritma konvergen ke solusi. Mari kita periksa skornya:

```python
print(f"Fitness solusi: {fitness(solution)}")
print(f"\nInterpretasi solusi:")
for i, color in enumerate(solution):
    print(f"Node {i}: warna {'Merah' if color == 'R' else 'Hijau'}")
```

Solusi memiliki skor 4, yang berarti optimal, karena kita memiliki tepat 4 edge dalam graf kita, artinya semua valid!

## 2. Masalah Delapan Ratu (Eight Queens Problem)

Mari kita lihat masalah yang lebih rumit.

### Deskripsi Masalah

Dalam masalah *Eight Queens*, kita ditugaskan untuk menempatkan delapan ratu di papan catur 8x8 tanpa ada ratu yang mengancam yang lain (alias ratu tidak boleh berada di baris, kolom, atau diagonal yang sama). Dalam bentuk umumnya, masalah didefinisikan sebagai menempatkan *N* ratu di papan catur NxN tanpa konflik.

### Representasi Solusi

Kita memiliki delapan ratu, jadi kita akan memiliki gen untuk masing-masing. Gene pool akan berupa angka dari 0 hingga 7, untuk kolom yang berbeda. *Posisi* gen dalam state akan menunjukkan baris di mana ratu tertentu ditempatkan.

Misalnya, kita dapat memiliki state "03304577". Di sini:
- Gen pertama dengan nilai 0 berarti "ratu di baris 0 ditempatkan di kolom 0"
- Gen kedua dengan nilai 3 berarti "ratu di baris 1 ditempatkan di kolom 3"
- Dan seterusnya...

### Fungsi Fitness

Sekarang kita perlu memikirkan fungsi fitness. Pada masalah pewarnaan graf kita menghitung edge yang valid. Proses pemikiran yang sama dapat diterapkan di sini. Alih-alih edge, kita memiliki posisi antar ratu. Jika dua ratu tidak saling mengancam, kita katakan mereka berada pada posisi "tidak menyerang". Kita dapat menghitung berapa banyak posisi seperti itu.

Mari kita langsung menginisialisasi populasi kita:

```python
population = init_population(100, range(8), 8)
print("5 individu pertama dari populasi:")
for i in range(5):
    print(f"Individu {i+1}: {population[i]}")
```

Kita memiliki populasi 100 dan setiap individu memiliki 8 gen. Gene pool adalah bilangan bulat dari 0 hingga 7.

Selanjutnya kita perlu menulis fungsi fitness kita. Ingat, ratu saling mengancam jika mereka berada di baris, kolom, atau diagonal yang sama.

Karena posisi bersifat mutual, kita harus berhati-hati untuk tidak menghitungnya dua kali. Oleh karena itu untuk setiap ratu, kita hanya akan memeriksa konflik untuk ratu setelahnya.

Nilai gen dalam individu `q` menunjukkan kolom ratu, dan posisi gen menunjukkan barisnya. Kita dapat memeriksa apakah nilai yang disebutkan di atas antara dua gen sama. Kita juga perlu memeriksa diagonal. Ratu *a* berada di diagonal ratu lain, *b*, jika perbedaan baris di antara mereka sama dengan perbedaan kolom mereka (untuk diagonal di kanan *a*) atau sama dengan negatif perbedaan kolom mereka (untuk diagonal kiri *a*).

Di bawah ini diberikan fungsi fitness:

```python
def fitness(q):
    """Menghitung jumlah pasangan ratu yang tidak saling menyerang."""
    non_attacking = 0
    for row1 in range(len(q)):
        for row2 in range(row1+1, len(q)):
            col1 = int(q[row1])
            col2 = int(q[row2])
            row_diff = row1 - row2
            col_diff = col1 - col2

            # Cek apakah tidak saling menyerang
            if col1 != col2 and row_diff != col_diff and row_diff != -col_diff:
                non_attacking += 1

    return non_attacking
```

Perhatikan bahwa skor terbaik yang dapat dicapai adalah 28. Itu karena untuk setiap ratu kita hanya memeriksa ratu setelahnya. Untuk ratu pertama kita memeriksa 7 ratu lain, untuk ratu kedua 6 lainnya dan seterusnya. Singkatnya, jumlah pemeriksaan yang kita lakukan adalah jumlah 7+6+5+...+1, yang sama dengan 7*(7+1)/2 = 28.
