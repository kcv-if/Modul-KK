Pokoknya, abis fuzzy setnya di agregasi (digabung) kita bikin region per regionnya (bikin banyak subdivisi) yang regionnya membentuk sebuah bidang segitiga/trapezoid gitulah. 

Berikutnya, kita komputasikan areanya yang dilingkupi dengan set fuzzy agregasi dengan menghitung setiap garis yang membentuk regionnya, menggunakan rumus persamaan garis lurus
 $$
\frac{y - y_1}{x - x_1} = \frac{y_2 - y_1}{x_2 - x_1}
$$

<b>Catatan</b> kalau kemiringannya (slope) = 0 (atau garisnya datar). nilai $y$ (derajat keanggotaan) tidak berubah (kan datar)

intinya kita mau menyimpan kemiringan dan jarak garis, biar mencari persamaan matematis untuk setiap segmen fungsi keanggotaan dari suatu himpunan fuzzy. kita lagi mendefinisikan

$$
\mu(x) =
\begin{cases}
0.5x, & x \in [0,1] \quad (\text{Segmen ab}) \\
0.5, & x \in [1,3.5] \quad (\text{Segmen bc}) \\
0.6x - 1.6, & x \in [3.5,4] \quad (\text{Segmen cd}) \\
\vdots & \text{lainnya}
\end{cases}
$$
